## Text Classification Task

```python
import torch
import torchtext
import os
import collections
```

### The Dataset
4 classes: World, Sports, Business, Sci/ Tech

```python

os.makedirs('./data', exist_ok=True)
train_dataset, test_dataset = torchtext.datasets.AG_NEWS(root = './data')
classes = ['World', 'Sports', 'Business', 'Sci/Tech']
```

```python
first_label, first_text = next(iter(train_dataset))
print(f"Original Label: {first_label} ({classes[first_label - 1]})")
print(f"Sample Text:   {first_text[:120]}...\n")
```

### Tokenization
Now, convert text into numbers that can be represent as tensors. If we want word-level representation, we need to do 2 things:
- Use **tokenizer** to split text into **tokens**.
- Build a **vocabulary** of those tokens.

```python
tokenizer = torchtext.data.utils.get_tokenizer('basic_english')
tokenizer('He said: Hello!')
```

```python
counter = collections.Counter()
for (label, line) in train_dataset:
    counter.update(tokenizer(line))

vocab = torchtext.vocab.vocab(counter, min_freq = 1)
```

```python
vocab_size = len(vocab)
print(f"Vocab size if {vocab_size}")

stoi = vocab.get_stoi()

def encode(x):
    return [stoi[s] for  s in tokenizer(x)]

encode('I love to play football')
```
